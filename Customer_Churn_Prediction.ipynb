```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# 1. Prepare features (X) and target (y)
X = df.drop(columns=["Churn", "customerID"])
y = (df["Churn"] == "Yes").astype(int)

# 2. Split the data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 3. Identify numerical and categorical columns
numeric_cols = X.select_dtypes(include="number").columns.tolist()
categorical_cols = X.select_dtypes(exclude="number").columns.tolist()

# 4. Prepare the data
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_cols),
    ("categorical", categorical_pipeline, categorical_cols)
])

# 5. Define new models
models = {
    "Decision Tree": DecisionTreeClassifier(
        max_depth=6, class_weight="balanced", random_state=42
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),
    "SVM": SVC(
        kernel="rbf", class_weight="balanced", probability=True,
        random_state=42
    ),
    "KNN": KNeighborsClassifier(n_neighbors=5)
}

# 6. Train and evaluate each model
results = []
trained_models = {}

for name, model in models.items():
    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)

    predictions = pipeline.predict(X_test)
    probabilities = pipeline.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1-score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

    trained_models[name] = pipeline

results_df = pd.DataFrame(results).set_index("Model")
results_df = results_df.sort_values("F1-score", ascending=False)

display(results_df.round(3))
```

Gini : mesure à quel point les classes sont mélangées dans le nœud. Une valeur proche de zéro indique qu'une classe domine.



Samples : nombre de clients d'entraînement qui arrivent à ce nœud.



Value : poids des observations des deux classes. Ici, les poids sont équilibrés à cause de class_weight="balanced".



Class : classe que le modèle prédit dans ce nœud, en fonction des poids des classes.

Decision Tree — résumé pour un entretien
Un arbre de décision est un algorithme supervisé de classification qui apprend des règles à partir des données d'entraînement pour prédire une classe. Il divise les données selon des critères qui améliorent la séparation entre les classes. Dans notre projet, il prédit si un client va quitter l'entreprise (Left) ou rester (Stayed).
Avantages : facile à interpréter et capable de modéliser des relations non linéaires.
Limites : risque de surapprentissage (overfitting) si l'arbre est trop complexe.
Notre résultat : accuracy de 74,0 % et recall de 78,9 % sur le jeu de test.

1. How does SVM work?
Imagine each customer as a point in a graph. Some customers left the company, while others stayed.
SVM tries to find a boundary that separates these two groups, ideally with the largest possible margin between the classes.

```python

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# All six models
all_models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42
    ),
    "Decision Tree": DecisionTreeClassifier(
        max_depth=6,
        class_weight="balanced",
        random_state=42
    ),
    "SVM": SVC(
        kernel="rbf",
        class_weight="balanced",
        probability=True,
        random_state=42
    ),
    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),
    "KNN": KNeighborsClassifier(n_neighbors=5)
}

# Train and evaluate every model
all_results = []
all_trained_models = {}

for name, model in all_models.items():
    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)

    pred = pipeline.predict(X_test)
    prob = pipeline.predict_proba(X_test)[:, 1]

    all_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(
            y_test, pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, pred, zero_division=0
        ),
        "F1-score": f1_score(
            y_test, pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, prob)
    })

    all_trained_models[name] = pipeline

comparison_df = (
    pd.DataFrame(all_results)
    .set_index("Model")
    .sort_values("F1-score", ascending=False)
)

display(comparison_df.round(3))
```

```python

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Build the Random Forest pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

# Hyperparameters to test
param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [10, 20, None],
    "model__min_samples_leaf": [1, 3]
}

# Cross-validation: preserve class proportions in each fold
cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

# Train and find the best settings
grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation F1-score:")
print(round(grid_search.best_score_, 3))
```

```python

from sklearn.metrics import classification_report, roc_auc_score

best_rf = grid_search.best_estimator_

rf_predictions = best_rf.predict(X_test)
rf_probabilities = best_rf.predict_proba(X_test)[:, 1]

print("Tuned Random Forest — Test Results")
print(classification_report(
    y_test,
    rf_predictions,
    target_names=["Stayed", "Left"]
))

print(
    "ROC-AUC:",
    round(roc_auc_score(y_test, rf_probabilities), 3)
)
```

```python

from sklearn.metrics import (
    accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score
)

# Evaluate the tuned Random Forest
tuned_result = {
    "Model": "Tuned Random Forest",
    "Accuracy": accuracy_score(y_test, rf_predictions),
    "Precision": precision_score(
        y_test, rf_predictions, zero_division=0
    ),
    "Recall": recall_score(
        y_test, rf_predictions, zero_division=0
    ),
    "F1-score": f1_score(
        y_test, rf_predictions, zero_division=0
    ),
    "ROC-AUC": roc_auc_score(y_test, rf_probabilities)
}

# Combine the original six models and tuned Random Forest
final_comparison = pd.concat([
    comparison_df,
    pd.DataFrame([tuned_result]).set_index("Model")
])

final_comparison = final_comparison.sort_values(
    "F1-score", ascending=False
)

display(final_comparison.round(3))
```

```python

%pip install -q plotly

import numpy as np
import pandas as pd
import plotly.graph_objects as go
from sklearn.metrics import accuracy_score

# =====================================================
# CHURN CITY — 3D CUSTOMER INTELLIGENCE
# =====================================================

# 1. Predict customer churn
predictions = best_rf.predict(X_test)

# Convert labels to 0 and 1
actual = np.asarray(y_test).astype(int)
predicted = np.asarray(predictions).astype(int)

# 2. Align test customers with original data
city = df.loc[X_test.index].copy().reset_index(drop=True)

city["Actual"] = actual
city["Predicted"] = predicted

# Identify prediction results
city["Result"] = np.select(
    [
        (city["Actual"] == 1) & (city["Predicted"] == 1),
        (city["Actual"] == 0) & (city["Predicted"] == 0),
        (city["Actual"] == 1) & (city["Predicted"] == 0),
    ],
    [
        "Correctly detected churn",
        "Correctly detected stay",
        "Missed churn"
    ],
    default="False alarm"
)

# 3. Create city blocks from contract and tenure
city["TenureGroup"] = pd.cut(
    city["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0–12 months", "13–24 months",
            "25–48 months", "49–72 months"]
)

city["Block"] = (
    city["Contract"].astype(str) + " | " +
    city["TenureGroup"].astype(str)
)

# Aggregate customer groups into city buildings
buildings = (
    city.groupby("Block", observed=True)
    .agg(
        Customers=("Churn", "size"),
        ChurnRate=("Actual", "mean"),
        PredictedChurn=("Predicted", "mean"),
        CorrectChurn=("Result",
            lambda s: (s == "Correctly detected churn").sum()),
        MissedChurn=("Result",
            lambda s: (s == "Missed churn").sum()),
        Contract=("Contract", "first"),
        TenureGroup=("TenureGroup", "first")
    )
    .reset_index()
)

# Height represents churn rate, with a minimum visible height
buildings["Height"] = 3 + buildings["ChurnRate"] * 25

# 4. Arrange buildings in a grid
n_cols = 5
buildings["X"] = np.arange(len(buildings)) % n_cols * 12
buildings["Y"] = np.arange(len(buildings)) // n_cols * 12

# 5. Create 3D city
fig = go.Figure()

# City ground
fig.add_trace(go.Mesh3d(
    x=[-8, 65, 65, -8],
    y=[-8, -8, 65, 65],
    z=[0, 0, 0, 0],
    i=[0, 0],
    j=[1, 2],
    k=[2, 3],
    color="#10172D",
    opacity=0.95,
    hoverinfo="skip",
    showlegend=False
))

# Neon street grid
for x in range(0, 61, 6):
    fig.add_trace(go.Scatter3d(
        x=[x, x], y=[-8, 65], z=[0.15, 0.15],
        mode="lines",
        line=dict(color="#27385A", width=2),
        hoverinfo="skip",
        showlegend=False
    ))

for y in range(0, 61, 6):
    fig.add_trace(go.Scatter3d(
        x=[-8, 65], y=[y, y], z=[0.15, 0.15],
        mode="lines",
        line=dict(color="#27385A", width=2),
        hoverinfo="skip",
        showlegend=False
    ))

# Colors represent model outcome for each customer group
colors = {
    "Correctly detected stay": "#20D9F3",
    "Correctly detected churn": "#FF3FA4",
    "Missed churn": "#FFB547",
    "False alarm": "#A78BFA"
}

# Each building is a 3D rectangular prism
for idx, row in buildings.iterrows():

    x, y, h = row["X"], row["Y"], row["Height"]
    w = 5.5
    color = (
        "#FF3FA4" if row["ChurnRate"] >= 0.4
        else "#20D9F3"
    )

    # Building walls and roof
    vertices = [
        (x, y, 0), (x+w, y, 0),
        (x+w, y+w, 0), (x, y+w, 0),
        (x, y, h), (x+w, y, h),
        (x+w, y+w, h), (x, y+w, h)
    ]

    vx, vy, vz = zip(*vertices)

    # Rectangular prism triangles
    i = [0,0,4,4,0,0,1,1,2,2,3,3]
    j = [1,2,5,6,1,5,2,6,3,7,0,4]
    k = [2,3,6,7,5,4,6,5,7,6,4,7]

    fig.add_trace(go.Mesh3d(
        x=vx, y=vy, z=vz,
        i=i, j=j, k=k,
        color=color,
        opacity=0.88,
        flatshading=True,
        name="Customer district",
        legendgroup="buildings",
        showlegend=(idx == 0),
        hovertemplate=(
            f"<b>{row['Block']}</b><br>"
            f"Customers: {row['Customers']}<br>"
            f"Actual churn rate: {row['ChurnRate']:.1%}<br>"
            f"Predicted churn rate: {row['PredictedChurn']:.1%}"
            "<extra></extra>"
        )
    ))

    # Glowing roof outline
    fig.add_trace(go.Scatter3d(
        x=[x, x+w, x+w, x, x],
        y=[y, y, y+w, y+w, y],
        z=[h+0.08]*5,
        mode="lines",
        line=dict(color="#FFFFFF", width=3),
        hoverinfo="skip",
        showlegend=False
    ))

    # Floating rooftop light
    fig.add_trace(go.Scatter3d(
        x=[x+w/2],
        y=[y+w/2],
        z=[h+0.5],
        mode="markers",
        marker=dict(
            size=4,
            color=color,
            line=dict(color="white", width=0.5)
        ),
        hoverinfo="skip",
        showlegend=False
    ))

# 6. Add stars above the city
rng = np.random.default_rng(42)
stars_x = rng.uniform(-8, 65, 180)
stars_y = rng.uniform(-8, 65, 180)
stars_z = rng.uniform(25, 42, 180)

fig.add_trace(go.Scatter3d(
    x=stars_x, y=stars_y, z=stars_z,
    mode="markers",
    marker=dict(
        size=rng.uniform(1, 2.5, 180),
        color="#FFFFFF",
        opacity=0.65
    ),
    hoverinfo="skip",
    showlegend=False
))

# 7. Dashboard style
fig.update_layout(
    title=dict(
        text=(
            "CHURN CITY"
            "<br><sup>AI CUSTOMER INTELLIGENCE | 3D METROPOLIS</sup>"
        ),
        x=0.5,
        font=dict(size=27, color="#F4F7FF")
    ),
    height=850,
    paper_bgcolor="#070B19",
    font=dict(color="#E7EEFF", family="Arial"),
    margin=dict(l=0, r=0, t=100, b=0),
    legend=dict(
        title="BUILDING COLOR",
        bgcolor="rgba(12,18,38,0.9)"
    ),
    scene=dict(
        bgcolor="#070B19",
        xaxis=dict(
            title="CITY BLOCK",
            backgroundcolor="#0B1225",
            gridcolor="#253451",
            showbackground=True
        ),
        yaxis=dict(
            title="CUSTOMER DISTRICT",
            backgroundcolor="#0B1225",
            gridcolor="#253451",
            showbackground=True
        ),
        zaxis=dict(
            title="CHURN RATE — BUILDING HEIGHT",
            backgroundcolor="#0B1225",
            gridcolor="#253451",
            showbackground=True
        ),
        camera=dict(
            eye=dict(x=1.7, y=1.7, z=1.35)
        ),
        aspectmode="manual",
        aspectratio=dict(x=1.3, y=1.3, z=0.85)
    )
)

fig.show(renderer="colab")
```

How to interpret the chart
- Accuracy: How often the model predicts correctly overall.
- Precision: When it predicts that a customer will leave, how often is it right?
- Recall: Of all customers who actually leave, how many does it detect?
- F1-score: A balance between precision and recall.
- ROC-AUC: How well the model distinguishes customers who leave from those who stay.
